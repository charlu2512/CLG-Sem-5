## Practical 4

Write Python programs using user-defined functions with different
argument types and demonstrate use of built-in modules (math,
random, statistics).

```python
import math
import random
import statistics
```

```python
# Function with no return type and no argument

def greet():
    print("Good morning !!!")

print(greet())
```

```python
# Function with no return type but with argument

def star_pattern(n):
    for i in range(0,n+1):
        print(" " * (n-i) + "* " * i)

        
star_pattern(5)
```

```python
#Function with return type and no argument

def dog_bark():
    return "woof woof !!!"

print("dog barks: " + dog_bark())
```

```python
# Function with return type and argument

def add(a,b):
    return(a+b)

a=[5,6,4,3,4,4]
b=[78,65,4,3,23]
x = add(a,b)
print(x)
```

```python
# Function with default argument

def greet(name="Alice"):
    print("Hello", name)

greet()
greet("Bob")
```

```python
#simple calculator

def calc(a,b):    
    return a+b,a-b,a*b,a/b,a%b

x=int(input("Enter number 1:"))
y=int(input("Enter number 2:"))

a,b,c,d,e = calc(x,y)

print("\nSimple calculator\n")
print("number 1: "+ str(x))
print("number 2: "+ str(y))
print("Addition: "+ str(a))
print("Subtraction: "+ str(b))
print("Multiplication: "+ str(c))
print("Division: "+ str(d))
print("Modulus: "+ str(e))
    
```

```python
#Math library in python

x=math.pi
print(x)
print("Square root:", math.sqrt(25))
print("Power:", math.pow(2, 3))
print("Ceiling:", math.ceil(4.3))
print("Floor:", math.floor(4.8))
p=math.sin(math.radians(30))
print(p)
```

```python
# random library in python

r1 = str(random.random())# returns random real number between 0 and 1
print("r1 : "+ r1)
r2 = str(random.randint(1,10))# returns random integer between 1 to 10
print("r2 : "+ r2)
```

```python
x = [1,2,3,4,5,6,7]
random.shuffle(x)
print(x)

i = random.choice(x)
print(i)
```

```python
#statistics library in python

numbers = [10, 20, 30, 40, 50]

print("Mean:", statistics.mean(numbers))
print("Median:", statistics.median(numbers))
print("Mode:", statistics.mode([10, 20, 20, 30, 40]))
```
