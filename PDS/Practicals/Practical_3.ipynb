## Practical 3

Implement programs using if, if-else, elif, while loop, for loop,
nested loops, and loop control statements (break, continue, pass).

```python
# if statement

number = 10

if number > 0:
    print("Number is positive")
```

```python
# if-else statement

number = 7

if number % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
# elif statement

marks = 75

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
```

```python
# while loop

i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
# for loop

for i in range(1, 6):
    print(i)
```

```python
# Nested loops

for i in range(1, 4):
    for j in range(1, 4):
        print(i, j)
```

```python
# break statement

for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

```python
# continue statement

for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```

```python
# pass statement

for i in range(1, 6):
    if i == 3:
        pass
    print(i)
```
