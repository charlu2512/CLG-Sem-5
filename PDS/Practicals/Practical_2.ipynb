```python
#SET

c={2,2,3,3,4,5,5,6,6,6,6}

print(f"List: {c}")


#String And numbers methods

txt = "Hello Charlie"
numbers = [2,3,2,3,4,6,5,8,9]


print(f"count:{txt.count('l')}")
print(f"index:{txt.index('C')}")


numbers.extend([55,65])
print(f"extend:{numbers}")

numbers_2 = numbers.copy()
print(f"numbers_2 (copy):{numbers_2}")

numbers_2.clear()
print(f"numbers_2 (clear):{numbers_2}")

numbers.insert(2,"Neo")
print(f"insert:{numbers}")

print(f"pop:{numbers.pop()}")

numbers.remove("Neo")
print(f"remove:{numbers}")

numbers.reverse()
print(f"reverse: {numbers}")

numbers.sort()
print(f"sort: {numbers}")
```
