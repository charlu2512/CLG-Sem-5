# Working with txt file in Python

```python
# opening prac5_1.txt file in write mode

f = open("prac5_1.txt","w")
f.write("Hello world!! This is a txt written in prac4.txt file using python")
f.close()
```

```python
# opening prac5_1.txt file in read mode

f = open("prac5_1.txt","r")
r = f.read()
print(r)
f.close()
```

```python
# rewriting the file again in write mode deletes all the previously stored txt data

f = open("prac5_1.txt","w")
f.write("Good Morning Everyone!!! ")
f.close()

# reading file

f = open("prac5_1.txt","r")
content = f.read()
print(content)
f.close()
```

```python
#opening prac5_1.txt file in append mode(it doesnt delete previously stored txt data)

f = open("prac5_1.txt","a")
f.write("\nThis line is the appended text. ")
# f.seek(0)
f.close()

f = open("prac5_1.txt","r")
print(f.read())

f.close()
```

```python
# opening prac5_1.txt file in r+ mode
# this mode is used to read and write already existing file. It does not create a new file if it does not exist.

f = open("prac5_1.txt","r+")
f.write("The pizza tastes amazing")
f.seek(0)# we must seek/move cursor to the beginning of file to print the whole data
print(f.read())
f.close()
```

```python
# opening prac5_2.txt file in w+ mode
# this mode is used to write and read w+ file. It  creates a new file if it does not exist.

f = open("prac5_2.txt","w+")
f.write("It takes 24 hrs to make sourdough pizza ")
f.write("\nItalians make it this way. ")
f.seek(0) # we must seek/move cursor to the beginning of file to print the whole data
print(f.read())
f.close()
```

```python
# opening prac5_2.txt file in a+ mode
# this mode is used to append and read a file. It creates a new file if it does not exist.

f = open("prac5_2.txt","a+")
f.write("\nIt is important to rinse our mouth after coming back home. ")
f.write("\nThis ensures we do not fall ill/have cold and reduces its chances. ")
f.seek(0) # we must seek/move cursor to the beginning of file to print the data
print(f.read())
f.close()
```

## Working with CSV file in python

```python
import csv
```

```python
#opening prac5_3,csv file in write mode

f = open("prac5_3.csv","w",newline = "")
writer = csv.writer(f) # writer object

#csv header
writer.writerow(["ID","name","discipline","hobby"])

#csv data
writer.writerow([1,"Alice","QA tester","Rubix cube solving"])
writer.writerow([2,"Bob","UI Design","pixel art"])
writer.writerow([3,"Charlie","Backend dev","Valorant"])
writer.writerow([4,"Neo","Cybersecurity","Anime"])

f.close()

f = open("prac5_3.csv","r",newline = "")

#reader object
reader = csv.reader(f)

#reading data
for row in reader:
    print(row)

f.close()
```

# Working with binary file in Python



```python
import pickle
```

```python
# Data to store
students = [
    [1, "Rahul", 88],
    [2, "Priya", 92],
    [3, "Amit", 85]
]

# Open binary file in write mode
f = open("prac5_4.dat", "wb")

# Write data to binary file
pickle.dump(students, f)

f.close()

print("Data written successfully!")

# Open binary file in read mode
f = open("prac5_4.dat", "rb")

# Read data from binary file
data = pickle.load(f)

print("\nReading data from binary file:")
for record in data:
    print(record)

f.close()
```
